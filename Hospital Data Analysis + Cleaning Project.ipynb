# Advanced Data Cleaning

```python
# Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
# standardize text
text_cols = ["Hospital_Name","Location","Department"]

for col in text_cols:
    df[col] = (
       df[col]
        .astype(str)
        .str.strip()
        .str.title()
)
df[text_cols].head(5) #Top 5 Recors and check to value fix and clean
```

```python
df
```

```python
# Text to Numeric Convert
numeric_cols = [
    "Doctors_Count",
    "Patients_Count",
    "Medical_Expenses"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# Covert into the date format
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"], dayfirst=True, errors="coerce")
df["Discharge_Date"] = pd.to_datetime(df["Discharge_Date"], dayfirst=True, errors="coerce")
```

```python
df
```

```python
# fix invalid Numbers in Numeric Columns
numeric_cols = [
    "Doctors_Count",
    "Patients_Count",
    "Medical_Expenses"
]

for col in numeric_cols:
    df.loc[df[col] < 0 , col] = np.nan

df[numeric_cols].head(5)
```

```python
df.info()
```

```python
df.dtypes
```

```python
# Date Column 

date_cols = ["Admission_Date", "Discharge_Date"]

# Step 1: Convert to datetime
for col in date_cols:
    df[col] = pd.to_datetime(df[col], dayfirst=True, errors="coerce")
```

```python
df.info()
```

```python
df
```

```python
# Feature Engineering
df["Stay_Days"] =  (df["Discharge_Date"] - df["Admission_Date"]).dt.days
```

```python
df
```

# 📈 PART 2: Visualization (Matplotlib)

```python
import matplotlib.pyplot as plt
```

```python
# 1️⃣ Total Number of Patients

total_patients = df['Patients_Count'].sum()
print(total_patients)
```

```python
plt.figure(figsize=(6,4))

bars = plt.bar(['Total_Patients'],[total_patients] ,color="yellow")
plt.title("Total Patients Across Hospitals")
plt.ylabel("Patients")

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,y,int(y),
             ha='center', va='bottom')

plt.show()
```

```python
# 2️⃣ Average Number of Doctors per Hospital
avg_doctors = df.groupby('Hospital_Name')['Doctors_Count'].mean().sort_values(ascending=False)

print(avg_doctors)
```

```python
colors = ["gold", "skyblue", "lightgreen", "salmon", "violet","red"]
plt.figure(figsize=(10,5))

ax = avg_doctors.sort_values().plot(kind='bar', color=colors[:len(avg_doctors)])
plt.title("Average Doctors Per Hospital")
plt.xlabel("Hospital")
plt.ylabel("Doctors")
plt.xticks(rotation=45)

# Show values on top of bars
for p in ax.patches:
    ax.annotate(round(p.get_height(),1),
               (p.get_x() + p.get_width()  / 2, p.get_height()),
                  ha='center', va='bottom', )
plt.show()
```

```python
# 3️⃣ Top 3 Departments with Highest Patients
dept_Patients = df.groupby('Department')['Patients_Count'].sum().sort_values(ascending=False)

top3 = dept_Patients.sort_values(ascending=False).head(3)

print(top3)
```

```python
plt.figure(figsize=(6,5))

top3.plot(kind='pie', autopct='%1.1f%%',)

plt.title("Top 3 Departments with Highest Patients")
plt.xlabel("Department")
plt.ylabel("Patients")

plt.show()
```

```python
# 4️⃣ Hospital with Maximum Medical Expenses
hospital_expenses = df.groupby('Hospital_Name')['Medical_Expenses'].sum()

max_hospital = hospital_expenses.idxmax()

print(max_hospital)
```

```python
colors = ["gold","skyblue","lightgreen","salmon","violet","orange","red","cyan","pink","lime"]

plt.figure(figsize=(10,7))

ax = hospital_expenses.sort_values().plot(
    kind='barh',
    color=colors[:len(hospital_expenses)]
)

plt.title("Total Medical Expenses by Hospital")
plt.xlabel("Medical Expenses")
plt.ylabel("Hospital Name")

# Show values on bars
for i, v in enumerate(hospital_expenses.sort_values()):
    plt.text(v, i, f"{v:,.0f}", va='center')

plt.show()
```

```python
# 5️⃣ Daily Average Medical Expenses

df['Daily_Expense'] = df['Medical_Expenses'] / df['Stay_Days']

daily_avg = df.groupby('Hospital_Name')['Daily_Expense'].mean().sort_values(ascending=False)

print(daily_avg)
```

```python
colors = ["gold", "skyblue", "lightgreen", "salmon", "violet"]

plt.figure(figsize=(10,7))

ax = daily_avg.sort_values().plot(
    kind='barh',
    color=colors[:len(hospital_expenses)]
)

plt.title("Average Daily Medical Expense per Hospital")
plt.xlabel("Hospital")
plt.ylabel("Daily Expense")

plt.show()
```

```python
# 6️⃣ Longest Hospital Stay

longest_stay = df.loc[df['Stay_Days'].idxmax()]
print(longest_stay)
```

```python
# 7️⃣ Total Patients Treated Per City
city_patients = df.groupby('Location')['Patients_Count'].sum().sort_values(ascending=False)

print(city_patients )
```

```python
colors = ["gold", "skyblue", "lightgreen", "salmon", "violet","red"]
plt.figure(figsize=(10,5))

ax = city_patients.sort_values().plot(kind='bar', color=colors[:len(city_patients)])
plt.title("Total Patients Treated per City")
plt.xlabel("City")
plt.ylabel("Patients")

plt.xticks(rotation=45)

# Show values on top of bars
for p in ax.patches:
    ax.annotate(round(p.get_height(),1),
               (p.get_x() + p.get_width()  / 2, p.get_height()),
                  ha='center', va='bottom', )
plt.show()
```

```python
# 8️⃣ Average Length of Stay Per Department
avg_stay = df.groupby('Department')['Stay_Days'].mean().sort_values(ascending=False)

print(avg_stay)
```

```python
plt.figure(figsize=(10,5))

plt.plot(avg_stay, color='blue', linestyle='--', marker='o')

plt.title("Average Length of Stay per Department")
plt.xlabel("Department")
plt.ylabel("Days")

plt.subplots_adjust(bottom=0.50, right=1.50)

plt.show()
```

```python
# 9️⃣ Department with Lowest Patients

least_dept = dept_Patients.idxmin()

print("Department with Lowest Patients:", least_dept)
```

```python
# 🔟 Monthly Medical Expenses Report
df['Month'] = df['Admission_Date'].dt.to_period('M')

monthly_expenses = df.groupby('Month')['Medical_Expenses'].sum()

print(monthly_expenses)
```

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_expenses.index.astype(str), monthly_expenses.values,
         linestyle='--', marker='o')

plt.title("Monthly Medical Expenses")
plt.xlabel("Month")
plt.ylabel("Expenses")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```
